```python
#### Installs, as necessary ####
# in termial:
# pip install tensorflow

#### Import packages ####
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.models import Sequential
from tensorflow.keras.optimizers import Adam
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
import io as io
import os as os
from PIL import Image
#import pipeline_utilities as p_util
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Activation, Dropout, BatchNormalization
```

```python
#### Set directories and instantiate data lists
##
train_dir = './data/train'
valid_dir = './data/valid'
test_dir = './data/test/'
##
x_train = []
y_train = []
x_valid = []
y_valid = []
x_test = []
y_test = []
```

```python
#### Load training dataset ####

##
for direct in os.listdir(train_dir):
        print("Loading training dataset {}".format(direct))
        for filename in os.listdir(os.path.join(train_dir,direct)):
            try:
                img_path = os.path.join(train_dir,direct,filename)
                #print(img_path)
                img = Image.open(img_path)
                img = img.resize((32,32))
                img = np.array(img)
                img = img/255
                x_train.append(img)
                if direct == 'wildfire':
                    y_train.append(1)
                else:
                    y_train.append(0)
            except:
                print("exception")
```

```python
#### Load validation dataset ####

##
for direct in os.listdir(valid_dir):
    print("Loading validation dataset {}".format(direct))
    for filename in os.listdir(os.path.join(valid_dir,direct)):
        img_path = os.path.join(valid_dir,direct,filename)
        #print(img_path)
        img = Image.open(img_path)
        img = img.resize((32,32))
        img = np.array(img)
        img = img/255
        x_valid.append(img)
        if direct == 'wildfire':
            y_valid.append(1)
        else:
            y_valid.append(0)
```

```python
#### Load test dataset
##
for direct in os.listdir(test_dir):
    print("Loading test dataset {}".format(direct))
    for filename in os.listdir(os.path.join(test_dir,direct)):
        try:
            img_path = os.path.join(test_dir,direct,filename)
            #print(img_path)
            img = Image.open(img_path)
            img = img.resize((32,32))
            img = np.array(img)
            img = img/255
            x_test.append(img)
            if direct == 'wildfire':
                y_test.append(1)
            else:
                y_test.append(0)
        except:
                print("exception")
```

```python
# convert to numpy arrays
x_train = np.array(x_train)
x_test = np.array(x_test)
x_valid = np.array(x_valid)
y_train = np.array(y_train)
y_test = np.array(y_test)
y_valid = np.array(y_valid)
x_train.shape
```

```python
####
x_train_flatten = x_train.reshape(x_train.shape[0], -1)
x_test_flatten = x_test.reshape(x_test.shape[0], -1)
x_valid_flatten = x_valid.reshape(x_valid.shape[0], -1)
print(x_train_flatten.shape)
print(x_test_flatten.shape)
print(x_valid_flatten.shape)
print(y_train.shape)
print(y_test.shape)
print(y_valid.shape)
```

```python
# Step 1: Build the CNN model
cnn_model = Sequential([
    
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    BatchNormalization(),
    MaxPooling2D((2, 2)),
    
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    Conv2D(32, (3, 3), activation='relu', padding='same'),
    
    Flatten(),
    
    Dense(128, activation='relu'),
    Dense(128, activation='relu'),
    BatchNormalization(),
    Dropout(0.5),
    
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    BatchNormalization(),
    
    Dense(1, activation='sigmoid')  # Output layer with number of classes
])

# Step 2: Compile the model
cnn_model.compile(optimizer='adam',
              loss='binary_crossentropy',  # Loss function for multi-class classification
              metrics=['accuracy'])

cnn_model.summary()
```

```python
import pandas as pd
import time

# Initialize an empty dataframe to store results
results = pd.DataFrame(columns=["Batch Size", "Time Elapsed (s)", "Accuracy", "Loss"])

# Parameters
batch_sizes = range(100, 5100, 100)
epochs = 1

# Loop through different batch sizes
for batch_size in batch_sizes:
    start_time = time.time()
    
    # Train the model with the current batch size
    history = cnn_model.fit(
        x_train, y_train,
        validation_data=(x_valid, y_valid),
        epochs=epochs,
        batch_size=batch_size,
        verbose=0  # Suppress detailed training logs for brevity
    )
    
    # Evaluate the model on the test set
    loss, accuracy = cnn_model.evaluate(x_test, y_test, verbose=0)
    
    # Time taken for the loop
    elapsed_time = time.time() - start_time

    # Print results
    print(f"Batch Size: {batch_size}, Time: {elapsed_time:.2f} sec, Accuracy: {accuracy:.4f}, Loss: {loss:.4f}")
    
    # Append results to the dataframe
    temp_data = {
        "Batch Size": batch_size,
        "Time Elapsed (s)": elapsed_time,
        "Accuracy": accuracy,
        "Loss": loss}

    #
    results.loc[len(results)] = temp_data
```

```python
import matplotlib.pyplot as plt

# Assuming `results` DataFrame exists and contains the data
plt.figure(figsize=(12, 6))

# Plotting each column
plt.plot(results["Batch Size"], results["Time Elapsed (s)"], label="Time Elapsed (s)", marker='o')

# Adding labels, legend, and title
plt.xlabel("Batch Size")
plt.ylabel("Metrics")
plt.title("Performance Metrics vs Batch Size")
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```
